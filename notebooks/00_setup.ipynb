1. Create the empty GitHub repo **`iomt-compression-research`** at github.com (Public, **no** README, **no** .gitignore — this notebook writes them).

mount Drive → create folders → set git identity → store PAT to Drive (getpass) → init repo + remote → write embedded files → first commit/push → self-test.
